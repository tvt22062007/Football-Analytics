```python
import pandas as pd

df = pd.read_csv("Players.csv")

print("The oldest player(s):")
print(df[df["Age"]==df["Age"].max()])

print("Top 3 assist providers:")
print(df.sort_values("Assists",ascending=False).head(3))

print("Players with at least 10 goals and under 25:")
print(df[(df["Age"]<25) & (df["Goals"]>=10)])
```
